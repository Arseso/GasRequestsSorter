```python
import re

def find_sign(text):
    pattern = r'С уважением,\s*([^\.]+)\.'
    match = re.search(pattern, text, re.IGNORECASE)
    if(match):
        return re.sub(r'\s+', ' ', match.group(1)).strip()
    else:
        return 'NO'
```

```python
data['spec'] = data.res.apply(find_sign)
```

```python
data.spec.value_counts().to_csv('spec.csv')
```
