```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('test.csv')
```

```python
data.spec = data.spec.apply(lambda x: " ".join(x.split(" ")[:2]))
```

```python
data.spec.value_counts()
```

```python
targets = pd.concat([data.spec.value_counts()[:9]
                     ], axis=0).keys()
targets
```

```python
data = data[data['spec'].isin(targets)]
data.spec.value_counts()
specs = {key : idx + 1
         for idx, key in enumerate(data.spec.value_counts().keys())
         }
```

```python
with open('targets.txt', "wb") as f:
    f.write(str(specs).encode('utf-8'))
```

```python
import json

with open('targets.txt', 'r', encoding='utf-8') as f:
    targets = json.loads(f.read().replace("\'", "\""))

print(targets)
```

```python
spec_onehot = pd.get_dummies(data=data.spec, prefix='spec')
spec_onehot.apply(lambda x: x.apply(lambda x: np.float64(x)))
```

```python
data["features"] = data.loc[:, ["subj_req", "req"]].apply(
    lambda row: f"{row.iloc[0]}\n{row.iloc[1]}", axis=1
    )
```

```python
dataset = pd.concat([data.features, spec_onehot], axis=1)
dataset.to_csv("dataset.csv")
```
