```python
import pandas as pd
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from transformers import AutoTokenizer
import numpy as np
from tqdm import tqdm
```

```python
import torch
import torch._dynamo

# Отключаем torch dynamo перед обучением
torch._dynamo.config.suppress_errors = True
# или полностью отключаем
torch._dynamo.config.disable = True
```

```python
torch.manual_seed(123)
```

Loading dataset

```python
TRAIN_PROP = 0.8
```

```python
data = pd.read_csv("dataset.csv").drop(columns=["Unnamed: 0"])

data.loc[:, 'spec_Амбросимов Алексей':'spec_Трофимов Александр'] = data \
    .loc[:, 'spec_Амбросимов Алексей':'spec_Трофимов Александр'] \
    .apply(lambda col: col.map({False: 0.0, True: 1.0}), axis=0)
```

```python
spec_columns = [col for col in data.columns if col.startswith('spec_')]
class_names = [col.replace('spec_', '').replace('_', ' ') for col in spec_columns]
one_hot_matrix = data[spec_columns].values
labels = np.argmax(one_hot_matrix, axis=1)
class_counts = np.sum(one_hot_matrix, axis=0)
print("Названия классов:", class_names)
print("Количество samples по классам:", class_counts)
```

```python
labels_tensor = torch.tensor(labels, dtype=torch.long)
class_counts_tensor = torch.tensor(class_counts, dtype=torch.float)
class_weights = 1.0 / class_counts_tensor
samples_weights = class_weights[labels_tensor]
sampler = WeightedRandomSampler(
    weights=samples_weights,
    num_samples=len(samples_weights),
    replacement=True
)
```

```python
data_train = data.sample(n=int(data.shape[0]*TRAIN_PROP), random_state=123)
data_valid = data.drop(index=data_train.index)

data_train.reset_index(inplace=True, drop=True)
data_valid.reset_index(inplace=True, drop=True)
```

```python
target_cols = data.loc[:, 'spec_Амбросимов Алексей':'spec_Трофимов Александр'].columns

for idx, row in data.iterrows():
    try:
        target_values = row[target_cols].values.astype(np.float32)
    except ValueError as e:
        print(f"Error in row {idx}: {e}")
        print(f"Problematic values: {row[target_cols]}")
        break
```

```python
class Dataset(torch.utils.data.Dataset):
    def __init__(self, data: pd.DataFrame, tokenizer: AutoTokenizer, context_l: int, pad_token = 2):
        super(Dataset, self).__init__()
        self.data = data
        self.tok = tokenizer
        self.context = context_l
        self.pad_token = pad_token

    def __getitem__(self, index):
        row = self.data.iloc[index]
        feature = row.loc['features']
        encoded = self.tok.encode(feature, return_tensors="pt").squeeze()
        target = row.loc['spec_Амбросимов Алексей':'spec_Шулубина Олеся'].values

        # Handle sequences longer than context length
        if len(encoded) > self.context:
            # Truncate to context length
            encoded = encoded[:self.context]
            padding_needed = 0
        else:
            # Calculate padding needed
            padding_needed = self.context - len(encoded)
        # Pad the sequence
        padded_encoded = nn.functional.pad(
            encoded, (0, padding_needed),
            value=self.pad_token
        )
        return padded_encoded, torch.tensor(list(target), dtype=torch.float32)

    def __len__(self):
        return self.data.shape[0]
```

Import model (ruGPT3 small)

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification

model_id = "ai-forever/rugpt3small_based_on_gpt2"

tokenizer = AutoTokenizer.from_pretrained(model_id)
# model = AutoModelForSequenceClassification.from_pretrained(model_id)
```

```python
# torch.save(model, 'rugpt3s.pth')
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
device
```

```python
model = torch.load("rugpt3s.pth", map_location= device, weights_only=False)
```

Tuning

```python
for param in model.parameters():
    param.requires_grad = False

model.score = nn.Linear(768, 9)
```

```python
model.score.requires_grad = True
model.transformer.ln_f.requires_grad = True
model.transformer.h[11].requires_grad = True
model.transformer.h[10].requires_grad = True
```

```python
model
```

DataLoaders

```python
ds_train = Dataset(data_train, tokenizer, 768)
ds_valid = Dataset(data_valid, tokenizer, 768)
```

```python
dl_train = DataLoader(ds_train, batch_size=32, sampler=sampler)
dl_valid = DataLoader(ds_train, batch_size=32, sampler=sampler)
```

Training

```python
model = model.to(device)
```

```python
EPOCHS = 20
TOTAL_BATCHES = len(dl_train)
```

```python
criterion = nn.CrossEntropyLoss(weight=class_weights)
optimizer = torch.optim.AdamW(model.parameters(), lr=0.0001, weight_decay=1e-5)
scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=5, gamma=0.1)

best_val_loss = float('inf')

pbar = tqdm(range(5), desc="Epoch")
for epoch in pbar:
    # Training
    model.train()
    train_losses = []
    for batch_idx, (data, targets) in enumerate(dl_train):
        data, targets = data.to(device), targets.to(device)
        optimizer.zero_grad()
        outputs = model(data)
        if hasattr(outputs, 'logits'):
            outputs = torch.softmax(outputs, )
            loss = criterion(outputs.logits, targets)
        else:
            loss = criterion(outputs, targets)
        loss.backward()
        optimizer.step()
        train_losses.append(loss.item())

    # Validating
    model.eval()
    valid_losses = []
    with torch.no_grad():
        for batch_idx, (data, targets) in enumerate(dl_valid):
            data, targets = data.to(device), targets.to(device)
            outputs = model(data)
            if hasattr(outputs, 'logits'):
                v_loss = criterion(outputs.logits, targets)
            else:
                v_loss = criterion(outputs, targets)
            valid_losses.append(v_loss.item())

    # Calculate epoch averages
    avg_train_loss = sum(train_losses) / len(train_losses)
    avg_val_loss = sum(valid_losses) / len(valid_losses)
    scheduler.step()
    pbar.set_postfix({
        'train_loss': f'{avg_train_loss:.4f}',
        'val_loss': f'{avg_val_loss:.4f}',
        'lr': f'{scheduler.get_last_lr()[0]:.2e}'
    })
    # Save best model
    if avg_val_loss < best_val_loss:
        best_val_loss = avg_val_loss
        torch.save(model, 'rugpt3s_tuned_best.pth')
```

```python
def categorical_metrices(model, valid_set: DataLoader, num_classes: int):
    model.eval()
    
    # Инициализация счетчиков для каждого класса
    tp = {cl: 0 for cl in range(num_classes)}
    tn = {cl: 0 for cl in range(num_classes)}
    fp = {cl: 0 for cl in range(num_classes)}
    fn = {cl: 0 for cl in range(num_classes)}
    
    with torch.no_grad():
        for batch in valid_set:
            features, targets = batch[0], batch[1]
            features = features.to(device)
            targets = targets.to(device)
            preds = model(features).logits
            predicted_classes = torch.argmax(preds, dim=0)
            for class_idx in range(num_classes):
                try:
                    pred_mask = (predicted_classes == class_idx)
                    target_mask = (targets == class_idx)
                    tp[class_idx] += (pred_mask & target_mask).sum().item()
                    tn[class_idx] += ((~pred_mask) & (~target_mask)).sum().item()
                    fp[class_idx] += (pred_mask & (~target_mask)).sum().item()
                    fn[class_idx] += ((~pred_mask) & target_mask).sum().item()
                except RuntimeError:
                    print(pred_mask)
                    print(target_mask)
    return tp, tn, fp, fn

categorical_metrices(model, dl_valid, num_classes=16)
```
